```python
import numpy as np #numerical operations
import pandas as pd #for analyzing data
import matplotlib.pyplot as plt #or plotting graphs
from sklearn.model_selection import train_test_split #for training and testing the model
from sklearn.linear_model import LinearRegression #for implementing linear regression
```

```python
import pandas as pd
data = pd.read_csv('home_dataset.csv')

house_sizes = data['HouseSize'].values
house_rent = data['HousePrice'].values
```

```python
# Visualize the data
plt.scatter(house_sizes, house_rent, marker ='o', color ='blue')
plt.title('House Prices vs. House Size')
plt.xlabel('House Size (sq.ft)')
plt.ylabel('House Price ($)')
plt.show()
```

```python
# Split the data into training and testing sets
x_train, x_test, y_train, y_test = train_test_split(house_sizes, house_rent, test_size=0.2, random_state=42)
```

```python
# Reshape the data for NumPy
x_train = x_train.reshape(-1, 1)
x_test = x_test.reshape(-1, 1)

# Create and train the model
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
# Predict prices for the test set
predictions = model.predict(x_test)

# Visualize the predictions
plt.scatter(x_test, y_test, marker='o', color='blue', label='Actual Prices')
plt.plot(x_test, predictions, color='red', linewidth=2, label='Predicted Prices')
plt.title('Property Price Prediction with Linear Regression')
plt.xlabel('House Size (sq.ft)')
plt.ylabel('House Price (millions $)')
plt.legend()
plt.show()
```
